
# Indiana Jones and the Last Crusade - A Text Adventure Game

**Setting the scene:** Welcome, Dr. Henry "Indiana" Jones Jr. Archaeologist, professor, and adventurer. Your father has vanished while searching for the Holy Grail, the cup of Jesus Christ, said to grant eternal life. Armed only with his journal of clues, you must descend into the flooded catacombs beneath Venice, find the tomb of a Crusader knight, and read the shield that reveals the location of the Grail's location in a temple in the Canyon of the Crescent Moon. Three deadly trials guard the chamber that holds the chalice. Survive them all, and you face the final test. Among many chalices, only one is the true Grail. Choose wrong, and you die. Choose right, however, and you save your father and yourself.

**How to play:** Read every scene carefully to pick up all the clues. When prompted, type your choice (either a letter, word, or number) and press Enter. If you complete all challenges successfully, you win the game. If choose incorrectly in any of the five stages, Indiana Jones dies and you can restart the game for another try.

**Known bugs:** Sometimes when the code is run, the first user input field is pinned right beneath the end of the code cell and not at the correct part of the story. Story still works, also inconsistent pattern. Restarting kernel and clearing previous cell outputs usually helps.

**Source:** Spielberg, S. (Director). (1989). *Indiana Jones and the Last Crusade* [Film]. Paramount Pictures.

```python
"""
Title: Indiana Jones and the Last Crusade - A Text Adventure Game
Published: May 11th, 2026
Author: Ben Kuehl
Source: Spielberg, S. (Director). (1989). Indiana Jones and the Last Crusade
[Film]. Paramount Pictures

"""

# IMPORTS FOR GAME SETUP
# Import time package to use time.sleep() function for dramatic pauses

import time


# DEFINE INTRO
def intro():
    """
    Entry point. Wraps the game in a while loop so the player can
    replay after winning or losing. catacombs() starts the chain that
    runs through every stage and ends with either win() or fail().
    """

# Print functions to print the title
# Input function to prompt user to start playing the game

    print("\n")
    print("       INDIANA JONES AND THE LAST CRUSADE")
    print("              A Text Adventure")
    print("_" * 50)
    
    input(prompt = "Press any key + Enter to start the game ")
    
    catacombs()
    
    
# DEFINE STAGE 1: "VENICE CATACOMBS"

def catacombs():
    """
    Stage 1: The player must identify Sir Richard's tomb among three sacrophargi
    to reveal the clues for the three trials that lie ahead.
    """
    
    print("\n")
    print("            STAGE 1: VENICE CATACOMBS")
    print("_" * 50)
    
# Print function to set the scene of the stage and list three options
# User will pick one by entering a letter

    print("""
Beneath the library of the San Bernabas lies the flooded crypt of
the Crusaders. three sacrophargi line the chamber, each with a
different emblem:
    
    a) A knight's helm with a Crusader's cross
    b) A laurel wreath of a Roman senator
    c) A serpent wrapped around a goblet
""")
    
    
# Indiana Jones mumbles words (repeats the clue needed to make the right choice)
    
    print('Sir Richard was a Knight of the First Crusade...\n')
        
        
# Try/Except function to account for different user inputs
# ".lower()" formats the answer as lowercas (if user types A instead of a)
# ".strip" cuts any blank space in the user's answer
    
    try:
        tomb_choice = input(prompt = "Which tomb do you investigate? (a/b/c): ").lower().strip()
    except:
        fail()


# Multiple lists created to allow for variation in user input
# Option a is the correct answer

    correct_options = ["a", "knight", "1"]
    laurel_options = ["b", "laurel", "2"]
    serpent_options = ["c", "serpent", "3"]
    all_valid = laurel_options + serpent_options + correct_options 
    
# added while loop to show error message and keep re-prompting for user input if the user does not make valid entry
# Alternative instead of losing

    while tomb_choice not in all_valid:
        print("\nI did not expect that input, please try again.\n")
        tomb_choice = input(prompt = "Which tomb do you investigate? (a/b/c): ").lower().strip()
    
    
# If/elif/else clause used to evaluate user input
# If checks whether the user's input is in the lsit of correct answer (allowing for different entry options)
# If yes, the game continues and the clues are revealed

    if tomb_choice in all_valid:
        if tomb_choice in correct_options:
            print("""
You brush away the dust. Mounted on the sacrophargus is a
Crusader's shield. You read the Latin inscription:
            
    "I, Sir Richard, Knight of the First Crusade, guard the path
        to the Grail. Three trials stand between the worthy and the cup.
            
    FIRST:  Only the pentinent man shall pass.
            The pentinent man kneels before God.
                   
    SECOND: Tread upon the letters that spell His name.
            In the tongue of Rome, the Lord begins not with J,
            but with I.
                    
    THIRD:  Only in the leap from the lion's head will he prove
            his worth.
                    
Beyond, the Grail awaits. The cup of a carpenter, humble and
unadorned, gives life eternal."
            
A map on the shield points east, to the Canyon of the Crescent Moon.
""")
            
            
# If the condition is met, the second stage is called after a 1-second pause

            time.sleep(1)
            breath_of_god()


# # The first elif condition checks different user inputs that resemble the second possible answer again using a list
# If the elif condition is met, the user selected a wrong option, the print function narrates a "death" scene
# Fail function is triggered after 1-second suspense pause created with time.sleep(1) and the player can start again

        elif tomb_choice in laurel_options:
            print("""
A Roman senator's tomb. Nothing to do with the Grail. Methane
gas escapes as you lift the lid. Your torch ignites it.
""")
        
            time.sleep(1)
            fail()
        
        
# The second elif condition checks different user inputs that resemble the remaining possible answer again using a list
# If the elif condition is met, the user selected a wrong option, the print function narrates a "death" scene
# Fail function is triggered after 1-second suspense pause created with time.sleep(1) and the player can start again

        else:
            print("""
The carved "serpent" is no carving. It is a real viper, and it
strikes before you can pull away.
""")
        
            time.sleep(1)
            fail()
        
    else:
        print("\An error occured.\n")
        time.sleep(1)
        fail()

        
        
# DEFINE STAGE 2: "BREATH OF GOD" (Trial 1)

def breath_of_god():
    """
    Stage 2: The first trial. The breath of god. The player must recall the clue from
    Sir Richard's shield ("only the pentinent man shall pass") and kneel
    to crawl beneath the whirling blades.
    """
    print("\n")
    print("            STAGE 2: THE BREATH OF GOD")
    print("_" * 50)
    
          
# Print function with text to set the scene of the stage and remind user of the three clues
# Present user with four options to choose from to pass trial 1

    print("""
You enter the temple. A corridor stretches ahead, and massive
bronze blades cut through the air at head height with a rhytmic WHOOSH.
""")
    print("Only the pentinent man shall pass")
    print('"The pentinent man kneels before God"\n')
    print("""What do you do?
    
    a) Time the blades and sprint through
    b) Walk through carefully, eyes on the blades
    c) Kneel low and crawl forward beneath the blades
    d) Turn back and find another way
""")

    
# Try/except function to account for different user inputs
# ".lower()" is used to format answer as lower case, ".strip()" to erase any unwanted spacing

    try:
        choice = input(prompt = "Your choice: ").lower().strip()
    except:
        fail()

    
# Multiple lists created to account for variations in user input for the same intended response
# Option c is the correct answer

    sprint_options = ["a", "sprint", "run", "1"]
    walk_options = ["b", "walk", "2"]
    correct_options = ["c", "kneel", "crawl", "3"]
    back_options = ["d", "back", "turn", "4"]
    all_valid = sprint_options + walk_options + correct_options + back_options
    
# added while loop to show error message and keep re-prompting for user input if the user does not make valid entry
# Alternative instead of losing

    while choice not in all_valid:
        print("\nI did not expect that input, please try again.\n")
        choice = input(prompt = "What do you do? ").lower().strip()

        
# Nested conditional (if within if)
# Outer checks input validity the inner checks the user's choice
# If the user choice is both in all_valid and correct_options a success message is displayed and the next stage is called

    if choice in all_valid:
        if choice in correct_options:
            print("""
You drop to your knees and crawl. The blades sweep above your
head. You emerge on the far side, unharmed.
""")
            time.sleep(1)
            word_of_god()

            
# Two inner elif functions to test whether user input is wrong answer a or b after passing as a valid answer
# Testing "in" condition to account for varying user input for same intended answer
# Inner else function as remaining possible user answer (d) after passing as valid answer
# All three wrong answers trigger defeat message and call the fail function. User can restart afterwards

        elif choice in sprint_options:
            print("\nYou run. The first blade misses by inches. The second does not.\n")
            time.sleep(1)
            fail()
        elif choice in walk_options:
            print("\nYou step forward, watching the blades... until one takes your head.\n")
            time.sleep(1)
            fail()

# remainig valid option is going back
        else:
            print("\nYou retreat. The temple seals behind you. The blades close in.\n")
            time.sleep(1)
            fail()
            
            
# Outer else function to close nested conditional
# This else clause should never hit, as either user input is valid (see loop above) or reprompted
# If reprompted tested again and then either if or elif will be true. This else is mainly a safety net

    else:
        print("\nAn error occured.\n")
        time.sleep(1)
        fail()


# DEFINE STAGE 3: WORD OF GOD (Trial 2)

def word_of_god():
    """
    Stage 3: The second trial. The word of god. The player must spell the name of the
    Lord in Latin (I-E-H-O-V-A-H) one tile at a time. One wrong letter
    and the player falls to their death. Special message for J/I mistake.
    """
    print("\n")
    print("            STAGE 3: THE WORD OF GOD")
    print("_" * 50)

    
# Print function with text to set the scene of the stage
# Print function to provide user with instructions how to pass Trial 2

    print("""
A vast circular chamber. The floor is a grid of stone tiles,
each carved with a single Latin letter. Below them... only darkness.
""")
    print('"Tread upon the letters that spell the name of God (Jehovah).')
    print(' In the tongue of Rome, the name begins with J, not with I."\n')
    print("Step on each letter in order. One wrong tile and you fall.\n")
          

# Create list correct_letters for the seven letter sequence of the Latin word for Jehovah
# This sequence is the solution to Trial 2

    correct_letters = ["I", "E", "H", "O", "V", "A", "H"]

    
# Create for loop to iterate through each letter the player must step on
# First iteration is position = 0 and letter = I, then position = 1 and letter = E and so on

    for position, letter in enumerate(correct_letters):
        
        
# Try/except function wrapped around user input as safety net to account for invalid inputs
# Input requires user to enter a letter. ".upper()" capitalizes the letter and strip clears blank space
# Step is defined as the position + 1 (1 added to zero indexing) to show user start at step 1 of 7
# Except used as safety nat triggers fail function and return is used to break out of the loop after the fail

        try:
            step = input(f"Step #{position + 1} of 7: ").upper().strip()
        except:
            fail()

        
# Nested conditional (if inside if or else)
# If statement checks whether the user input (defined as step) is equal to the letter the loop is currently on
# If yes, "the tile holds firm" is printed and the loop continues to the next step/letter combination

        if step == letter:
            print("The tile holds firm.")

            
# If the if condition is not met, the else condition (wrong user inputs) distinguishes between two mistakes
# If the user put "J" instead of "I" on the first step, a special fail message is displayed providing a hint
# If the user makes a mistake on any other step, a standard fail message is displayed
# For all mistakes, the fail function is triggered after a time sleep and the loop is escaped with the return function

        else:
            if step == "J" and letter == "I":
                print("\nThe J tile crumbles. In Latin, the name begins with I, not J.\n")
                time.sleep(1)
                fail()
                return
            else:
                print("\nThe tile crumbles beneath your boot. You fall into darkness.\n")
                time.sleep(1)
                fail()
                return

        
# If the for loop completes, every letter was correct and the player advances
# The next stage is called after a time sleep

    print("\nThe far wall slides open with a grinding rumble.\n")
    time.sleep(1)
    path_of_god()


# DEFINE STAGE 4: PATH OF GOD (Trial 3)

def path_of_god():
    """
    Stage 4: The third trial. The leap of faith. The player must
    trust Sir Richard's clue and step off the lion's head into what
    appears to be a bottomless chasm. An invisible bridge will catch
    them, if they have the courage to take the step.
    """
    print("\n")
    print("            STAGE 4: THE PATH OF GOD")
    print("_" * 50)
    
    
# Print function with text to set the scene of the stage and remind user of the three clue
# Present user with four options to choose from to pass trial 3
   
    print("""
The floor ends at a vast chasm. Across the gap, you see the
Grail Chamber. There is no bridge. Only a stone lion's head
carved into the cliff edge. Behind you, the walls begin to grind shut.
""")
    print("Only in the leap from the lions head will he prove his worth.\n")
    print(""" You have only moments to decide.
    
   a) Search the walls for a hidden mechanism
   b) Throw a stone across to test if something is there
   c) Turn back and try to outrun the closing walls
   d) Step off the lion's head into the chasm
""")


# Try/except function to account for different user inputs
# The try block tries to understand the user input as choice between option a,b,c,d
# ".lower()" is used to format answer as lower case, ".strip()" to erase any unwanted spacing

    try:
        choice = input(prompt = "Your choice: ").lower().strip()
    except:
        fail()

    
# Multiple lists created to account for variations in user input for the same intended response
# Option d is the correct answer
 
    search_options = ["a", "search", "mechanism", "1"]
    throw_options = ["b", "throw", "stone", "2"]
    back_options = ["c", "back", "turn", "outrun", "3"]
    correct_options = ["d", "leap", "step", "jump", "4"]
    all_valid = search_options + throw_options + back_options + correct_options

    
# added while loop to show error message and keep re-prompting for user input if the user does not make valid entry
# Alternative instead of losing

    while choice not in all_valid:
        print("\nI did not expect that input, please try again.\n")
        choice = input(prompt = "Your choice: ").lower().strip()
    
# Nested conditional (if within if)
# Outer checks input validity the inner checks the user's choice
# If the user choice is both in all_valid and correct_options a success message is displayed and the next stage is called
    
    if choice in all_valid:
        if choice in correct_options:
            print("""
You close your eyes and step into the void. Your foot lands on
solid stone. An invisible bridge, painted to match the chasm
below. You walk across to the Grail Chamber.
""")
            time.sleep(1)
            grail_chamber()
            
            
# Two inner elif functions to test whether user input is wrong answer a or b after passing as a valid answer
# Testing "in" condition to account for varying user input for same intended answer
# Inner else function as remaining possible user answer (c) after passing as valid answer
# All three wrong answers trigger defeat message and call the fail function. User can restart afterwards

        elif choice in search_options:
            print("\nThere is no lever. The walls close. It is too late.\n")
            time.sleep(1)
            fail()
        elif choice in throw_options:
            print("\nThe stone vanishes silently into the chasm. The walls close around you.\n")
            time.sleep(1)
            fail()
        else:
            print("\nThe temple has already sealed itself. The walls grind togeter.\n")
            time.sleep(1)
            fail()
            
            
# Outer else function to close nested conditional
# This else clause should never hit, as either user input is valid (see loop above) or reprompted
# If reprompted tested again and then either if or elif will be true. This else is mainly a safety net

    else:
        print("\nAn error occured.\n")
        time.sleep(1)
        fail()


# DEFINE STAGE 5: THE GRAIL CHAMBER

def grail_chamber():
    """
    Stage 5: The final test. AMong five chalices, only the simple
    wooden cup of a carpenter is the true Grail. All others bring
    death.
    """
    print("\n")
    print("            STAGE 5: THE GRAIL CHAMBER")
    print("_" * 50)
    
    
# Print function with text to set the scene of the stage and remind user of the three clues   
    
    print("""
A stone table holds 5 chalices. At the back, the Grail Knight
watches in silence, kept alive seven hundred years by the cup
he guards.
""")
    print('" The cup of a carpenter, humble and unadorned, gives life eternal."\n')
    

# Create list of all the chalices on the table (all available options the user can pick from)
    
    chalices = [
        "A jewel-encrusted gold chalice",
        "A polished silver goblet",
        "A simple cup of plain, dark wood",
        "A cup of clear crystal, etched with stars",
        "A cup of beaten copper",
    ]

    
# Store "A simple cup of plain, dark wood" as the correct cup for readability
# Later referenced in conditional statement to trigger win function

    correct_cup = 3
    
    print("Before you stand five cups:\n")

    
# Use a for loop to display each chalice with its corresponding number
# start = 1 is used to start the list displayed to the user with 1 instead of 0 (zero-indexing)
# the f-string in the print function ensures we print the number and item itself

    for i, cup in enumerate(chalices, start=1):
        print(f"   {i}  {cup}")

    print("\nDrink from the wrong one and you will not leave alive.\n")

    
# Use try/except to account for non-numeric user input
# Input prompt used to ask user to enter a number from 1-5
# Wrap input function in an int() function to turn user input into an integer
# Use first except function if user input can't be transforemed to integer (e.g. abc) and call fail function
# Use second except function as safety net in case of other input related errors to end the game

    try:
        cup_choice = int(input(prompt = "Which cup do you drink from? (1-5): ").strip())
    except ValueError:
        print("\nYour hand trembles. Indecision is its own poison.\n")
        time.sleep(1)
        fail()
        return
    except:
        fail()

    
# If statement checks whether the user picked the right cup. If yes, win function is called, player wins
# If the cup is wrong, the else uses another if to check whether the cup choice was still a valid number
# If yes, the fail message is printed and the fail function is called
# If the user input is out of range, the fail message specifies this and then the fail function is called
# Either way the player can play again

    if cup_choice == correct_cup:
        win()
    else:
        if 1 <= cup_choice <= len(chalices):
            print(f"\nYou lift the {chalices[cup_choice - 1].lower()}. You drink.")
            print("The water turns to ash in your mouth. You crumble to dust.\n")
        else:
            print("\nYou reach for a cup that isn't there. The chamber goes dark.\n")
        time.sleep(1)
        fail()

        
#  DEFINE WIN AND FAIL FUNCTION

# Definition of win function. Only called if player successfully passes through all 5 stages.
# Added ASCII Art piece to enhance user experience through visuals
# Game ends with victory message and speach by Indiana Jone's father and Knight to help set the scene
# Used print statements for visual and speech

def win():
    """Define and display winning message and Chalice symbolizing the Holy Grail."""
    print("\n")
    print("          YOU CHOSE WISELY" + "\n")
    print("""@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@
@ : : : : : : : : : : : : : : : @
 @@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@
 |                             |
 |  \|/        %%%        \|/  |
 |  -t-     %%%%%%%%%     -t-  |
 |  /|\     \  %%%  /     /|\  |
 |         \ / %%% \ /         |
 \        - |  %%%  | -        /
  \       - |  %%%  | -       /
   \       / \     / \       /
    \        / --- \        /
     \         ! !         /
      \ __ ___ __ ___ ___ /
      ( ___ ___ __ _ ___  )
       (88888888888888888)
        --\ --------- /--
          ((((((o))))))
           \         /
            | | | | |
            | | | | |
            | | | | |
            | | | | |
            | | | | |
            | | | | |
            | | | | |
            | | | | |
          _(IIIIIIIII)_
       __/_____________\__
  ____/___________________\____
 /_____________________________\
""" + "\n")
    print("_" * 50)
    print("""
You lift the wooden cup. You fill it from the spring. You drink.
The water is cold and pure. Eternal life flows through you.

You have found...

The Holy Grail.
""")
    time.sleep(1)
    print('Dad (echoing): "Junior... you did it."')
    print('Grail Knight:  "You have chosen... wisely."\n')
    print("Congratulations, Dr. Jones. You have won\n")
    
    replay_option = input(prompt = "Want to play again? (y/n): ").lower().strip()
    
    if replay_option in ["y", "yes", "YES", "Yes", "1"]:
        intro()
    else:
        print("\nSafe travels, Dr. Jones")
                          
                          

# Definition of fail function. Can be called in each stage, whenever the player enters a wrong answer
# Added ASCII Art piece to enhance user experience through visuals
# Game ends with defeat message and speach by Knight to make scene more dramatic
# Used print statements for visual and speech


def fail():
    """Display the failure message and skull symbolizing death."""
    print("\n")
    print("                   YOU CHOSE POORLY" + "\n")
    print("""     .... NO! ...                  ... MNO! ...
   ..... MNO!! ...................... MNNOO! ...
 ..... MMNO! ......................... MNNOO!! .
.... MNOONNOO!   MMMMMMMMMMPPPOII!   MNNO!!!! .
 ... !O! NNO! MMMMMMMMMMMMMPPPOOOII!! NO! ....
    ...... ! MMMMMMMMMMMMMPPPPOOOOIII! ! ...
   ........ MMMMMMMMMMMMPPPPPOOOOOOII!! .....
   ........ MMMMMOOOOOOPPPPPPPPOOOOMII! ...
    ....... MMMMM..    OPPMMP    .,OMI! ....
     ...... MMMM::   o.,OPMP,.o   ::I!! ...
         .... NNM:::.,,OOPM!P,.::::!! ....
          .. MMNNNNNOOOOPMO!!IIPPO!!O! .....
         ... MMMMMNNNNOO:!!:!!IPPPPOO! ....
           .. MMMMMNNOOMMNNIIIPPPOO!! ......
          ...... MMMONNMMNNNIIIOO!..........
       ....... MN MOMMMNNNIIIIIO! OO ..........
    ......... MNO! IiiiiiiiiiiiI OOOO ...........
  ...... NNN.MNO! . O!!!!!!!!!O . OONO NO! ........
   .... MNNNNNO! ...OOOOOOOOOOO .  MMNNON!........
   ...... MNNNNO! .. PPPPPPPPP .. MMNON!........
      ...... OO! ................. ON! .......
         ................................                   """ + "\n")
    print("_" * 50)
    print("""
The dust settles. A knight in tarnished armor lowers his head
and whispers: "He chose poorly."
""")
    replay_option = input(prompt = "Want to play again? (y/n): ").lower().strip()
    
    if replay_option in ["y", "yes", "YES", "Yes", "1"]:
        intro()
    else:
        print("\nSafe travels, Dr. Jones")
    
intro()    
```
